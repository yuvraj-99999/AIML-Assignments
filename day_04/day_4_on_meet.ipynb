try: Python attempts to run the code inside this block normally. If no error occurs, the except block is skipped.

except: If an error happens inside the try block, execution jumps here immediately instead of crashing and then the program continues running afterward.

```python
def div(a,b):
  try:
    return a/b
  except ZeroDivisionError:
    print("You cannot divide by zero")

num1 = int(input("Enter the first number:"))
num2 = int(input("enter the second number:"))

print(div(num1,num2))

```

```python
#zigzag program

import sys
import time

indent = 0  # How many spaces to indent
indent_increasing = True  # Whether the indentation is increasing or decreasing

try:
    while True:  # The main program loop
        print(" " * indent, end="")
        print("********")
        time.sleep(0.1)  # Pause for 1/10th of a second

        if indent_increasing:
            # Increase the number of spaces:
            indent += 1
            if indent == 20:
                # Change direction:
                indent_increasing = False
        else:
            # Decrease the number of spaces:
            indent -= 1
            if indent == 0:
                # Change direction:
                indent_increasing = True

except KeyboardInterrupt:
    # Handles Ctrl+C to exit
    print("\nProgram stopped.")
    sys.exit()
```
