```python
#0ZeroDivisionError Demonstration

def zeroDivErr(a,b):
  try:
    return a/b
  except ZeroDivisionError:
    print("You cannot divide by zero")

num1 = int(input("Enter the first number:"))
num2 = int(input("enter the second number:"))

print(zeroDivErr(num1,num2))
```

```python
#ValueError exception demonstration

def valErr(user_input):
    try:
        number = int(user_input)
        print(f"Success: {number}")
    except ValueError:
        print("ValueError: Cannot convert text to an integer.")

userIp = input("Enter a number: ")
valErr(userIp)
```

```python
#TypeError exception demonstration

def typeErr(a, b):
    try:
        result = a + b
        print(f"Result: {result}")
    except TypeError:
        print("TypeError: Cannot add a string and an integer together.")

typeErr('2', 2)
```

```python
#IndexError exception demonstration

def indexErr(items_list, index):
    try:
        item = items_list[index]
        print(f"Found item: {item}")
    except IndexError:
        print("IndexError: Requested list index is out of range.")

indexErr([1, 2], 5)
```
