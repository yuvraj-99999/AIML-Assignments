```python
class iterDemo:

  def __init__(self, data):
    self.data = data
    self.index = 0

  def __iter__(self):
    return self

  def __next__(self):
    if self.index >= len(self.data):
      raise StopIteration

    item = self.data[self.index]
    self.index+=1
    return item

element_list = ["Mr","Ryan","Gosling"]

for ele in iterDemo(element_list):
  print(ele)
```

```python
#decorator with the @decorator syntax

def decorator(func): #the parameter is another function
  def wrapper():
    print("Who is this?") #something to execute before our function
    func()
  return wrapper


@decorator
def who():
  identity = input("Who are you ?")
  print(identity)

who()
```

```python
#context managers

file = open("data.txt", "w")
file.write("Hello, Ryan Gosling this side.")
file.close()
```

```python
#without the context manager

file = open("data.txt", "r")

content = file.read()

print(content)

file.close()  # manually close the file
```

```python
#with the context manager

with open("data.txt", "r") as file:
  content = file.read()
  print(content) #the file is automatically closed after this block, no need to do file.close() manually
```

Manual file handling: You have to explicitly close the file using file.close().

Context manager (with): Python automatically closes the file after the with block, even if an error occurs.

```python
#creating our own context class

class MyContext:
    def __enter__(self):
        print("Entering")
        return self

    def __exit__(self, exc_type, exc_value, tb):
        print("Leaving")


with MyContext(): #for a class to work with 'with' it needs to provide two methods__enter__ and __exit__
    print("Inside")
```

```python
#demonstrating the exc_type,exc_value,tb

class MyContext:
    def __enter__(self):
        print("Entering")
        return self

    def __exit__(self, exc_type, exc_value, tb):
        print("Leaving")
        print("Exception Type:", exc_type)
        print("Exception Value:", exc_value)
        print("Traceback:", tb)


with MyContext():
    print("Inside")
    result = 2 +"2"
```

```python
#mini project

import time
from contextlib import contextmanager

#contextmanager is a decorator provided by Python that lets us create a context manager using a generator function instead of writing a class.


@contextmanager
def timer():

    # Record the starting time
    start = time.time()

    try:
        yield #yeild will basically make this block stop and execute the code within the with block

    finally: #we need this block to execute even if an exception occurs or code crashes
        print(f"Time: {time.time() - start:.4f}s") #the timer will resume and exceution time will be calculated

def train_one_epoch():
    time.sleep(2) #put to sleep for 2 seconds for just demonstrating


with timer():
    train_one_epoch()
```
