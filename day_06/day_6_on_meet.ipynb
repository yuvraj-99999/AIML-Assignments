```python
class Countdown:

    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < 0: #raises exception the value is lesser than 0
            raise StopIteration
        number = self.current
        self.current -= 1 #decrements the value of current value by 1
        return number


for number in Countdown(3): # calls the init method and start becomes 3 as well as current(instance variable)
    print(number)
```

```python
def get_numbers():
  yield 10
  yield 20
  yield 30

next(get_numbers())
next(get_numbers()) #this wont print 20 as this is generator and it will always start

```

```python
#this is the solution to the problem that occured above

def get_numbers():
  yield 10
  yield 20
  yield 30

numbers = get_numbers()
print(next(numbers))
print(next(numbers))
print(next(numbers))
```

```python
def count_up_to(limit):
    number = 1
    while number <= limit:
        yield number
        number += 1


for number in count_up_to(5): #basically before the loops starts the function has all the numbers the for loop just prints one by one
    print(number)
```

```python
#decorator without the @decorator syntax

def decorator(func): #the parameter is another function
  def wrapper():
    print("Who is this?") #something to execute before our function
    func()
  return wrapper


def say_hello():
  identity = input("Who are you ?")
  print(f"Hello {identity}")


say_hello = decorator(say_hello) #passing a function as an argument
say_hello()
```

```python
def decorator(function):
    def wrapper(*args, **kwargs):
        result = function(*args, **kwargs) #calling of original function
        return result

    return wrapper


@decorator #wraps add() with the decorator
def add(a, b):
    return a + b


add(10, 20) #calls the wrapper
```

```python
def decorator(function):
    def wrapper(*args, **kwargs):
        print("Executing function...") #print before calling the function
        result = function(*args, **kwargs)
        return result

    return wrapper


@decorator
def add(a, b, extra=0): #default value assigned to extra
    return a + b + extra


print(add(10, 20, extra=5)) #5 will overwrite 0 in extra
```
