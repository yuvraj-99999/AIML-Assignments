```python
import numpy as np

marks = np.array([90,70,85,99,100]) #creating the array
print(f"marks: {marks}")

#inbuilt functions of numpy

print(f"Sum of Marks: {np.sum(marks)}")
print(f"Average of Marks: {np.mean(marks)}")
print(f"Max of Marks: {np.max(marks)}")
```

```python
import requests

# Send an HTTP GET request to the specified website
response = requests.get("https://goslings.online")

# Print the HTTP status code (e.g., 200 for OK)
print(response.status_code)

# Print the raw content/HTML body of the response
print(response.text)

# Prints the method reference itself; to execute it and raise an exception if status is 4xx/5xx, call response.raise_for_status()
print(response.raise_for_status)
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title": "I AM BATMAN",
    "body": "Gotham needs me!",
    "userId": 1
}

response = requests.post(url, json=data)

print(response.status_code)
print(response.json())
```
