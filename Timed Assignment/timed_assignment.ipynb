```python
#Question 1

numbers = [12, 7, 9, 4, 22, 15, 8, 3, 30, 11]
oddCount = 0
evenCount = 0
evenList = []

for number in numbers:
  if(number % 2 == 0):
    print(number, "is even")
    evenList.append(number)
    evenCount += 1
  else:
    print(number, "is odd")
    oddCount += 1

print("Odd count is", oddCount)
print("Even count is", evenCount)
print("The Even List :",evenList)
```

```python
#Question 2

def calculate_average(numbers: list[int]) -> float:
    return sum(numbers) / len(numbers)


#normal list

marks: list[int] = [100,99,95,98,89,97]
average: float = calculate_average(marks)
print(f"The average of the given marks is {average}")



# Test with an empty list
try:
    result = calculate_average([])
    print(f"The average of the given marks is {average}")
except ZeroDivisionError:
    print("Cannot calculate the average of an empty list.")
```

```python
#Question 3

import random

guesses = []

for i in range(10):
  guess = random.randint(1,50)
  guesses.append(guess)

print("The Guesses are:", guesses)
print("The Highest Guess: ", max(guesses))
print(f"Average guess: {sum(guesses) / len(guesses):.2f}")
```

```python
#Question 4

class Student:


  def __init__(self, name: str, marks: list):

    """This is a special method which will invoke on object creation"""
    self.name = name
    self.marks = marks

  def average(self) -> float:
    """This method will calculate the average of the marks"""

    return sum(self.marks) / len(self.marks)

  def __str__(self):

    """This too is a special method which will invoke on object creation and will return a human readable string"""

    if(self.name == "Bruce Wayne"):
       return f"My name is {self.name} and by the way i am not BATMAN and the average of my marks is {self.average()}"
    return f"My name is {self.name} and the average of my marks is {self.average()}"

student1 = Student("Ryan Gosling", [100,90,80,99,95])

student2 = Student("Batman", [98,99,100,85,89])

student3 = Student("Bruce Wayne", [99,93,91,92,94])

students = [student1,student2,student3]

averageList: list =[]

for student in students:
  print(student)
  averageList.append(student.average())

maxAverage = max(averageList)

for student in students:
  if(student.average() == maxAverage):
    print(f"I am {student.name} amd i have the highest average: {student.average()}")
```

```python
#Q5- I build a random movie picker


import random

print("Confused about which Movie to watch today ?")
print("No worries we will pick the movie for you!!!")
num_of_movies:int = int(input(("Enter the number of movies you want to decide from:")))

movies: list[str] = []

for i in range(num_of_movies):
  movie = input(f"Enter Movie {i+1}: ")
  movies.append(movie)

random.shuffle(movies)

todays_movie = random.choice(movies)

print(f"Watch {todays_movie} today.... Enjoy!!!")
```
