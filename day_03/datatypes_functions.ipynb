```python
# String
text = "Ryan Gosling is literally me"

print(text.upper())
print(text.lower())
print(text.replace("literally", "actually"))
print(text.split())
print(text.find("Gosling"))


```

```python
# List

movies = ["Drive", "Barbie", "Blade Runner 2049", "The Nice Guys"]

movies.append("La La Land")
movies.insert(1, "The Fall Guy")
movies.remove("Barbie")
print(movies.pop())
movies.sort()
print(movies)
```

```python
# Tuple

ratings = (10, 8, 9, 10, 7)

print(ratings.count(10))
print(ratings.index(9))
print(len(ratings))
print(max(ratings))
print(min(ratings))
```

```python
# Dictionary

person = {
    "name": "Ryan Gosling",
    "role": "literally me",
    "movie": "Drive"
}

print(person.keys())
print(person.values())
print(person.items())
print(person.get("movie"))

person.update({"rating": 10})
print(person)
```

```python
# Set

characters = {"Ryan Gosling", "Ken", "Driver"}

characters.add("Sebastian")
characters.remove("Ken")

print(characters | {"Mysterious Guy"})
print(characters & {"Ryan Gosling", "Driver"})
print(characters - {"Driver"})
```

```python
# Integer and Float

x = -42
y = 3.14159

print(abs(x))
print(float(x))
print(int(y))
print(pow(2, 5))
print(round(y, 2))
```

```python
# Boolean

a = True
b = False

print(bool(1))
print(bool(0))
print(not a)
print(a and b)
print(a or b)
```

```python
# None

value = None

print(type(value))
print(value is None)
print(bool(value))
print(str(value))
print(value == None)
```
