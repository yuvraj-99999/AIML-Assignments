```python
name = ''
while name != 'your name':
    print('Please type your name.')
    name = input('> ')
print('Thank you!')
```

```python
#break stmt

while True:
    print('Type your name.')
    name = input('> ')
    if name == 'your name':
        break
print('Thank you!')
```

```python
#continue stmt

for i in range(10):
  if i%2 == 0:
    continue
  print(i)
```

break stops the entire loop completely and jumps straight out to the code below it.

continue skips only the rest of the current iteration and jumps straight to the next loop cycle.

```python
#sys code

import sys

while True:
  print("Type to exit")
  response = input(">")
  if response == 'exit':
    sys.exit()
  print("You typed", response)
```

sys.exit() terminates Python by throwing a SystemExit exception.

```python
import random

answer = random.randint(1,10)
attempts = 0
max_attempts = 5

print("Gosling Guess")
print(f"I am going to think of a number between 1 to 10 you have to guess it within  {max_attempts}")

while True:
  print("Enter Your Guess: ")
  number = int(input(">"))
  attempts+=1

  if number < answer:
    print("Too Low")
  elif number < answer:
    print("Too High")
  else:
    print(f"Congrats you guessed it within {attempts} attempts")
    break

  if(attempts == max_attempts):
    print("You ran out of attempts")
    print("The number was", answer)
```
